```python
import ollama

MODEL = "tinyllama"  # nome do modelo baixado
SYSTEM = (
    "Você é um agente pedagógico offline. "
    "Responda sempre em português brasileiro com clareza e objetividade."
)

def gerar(resumo_da_tarefa, entrada, n_tokens=512):
    messages = [
        {"role": "system", "content": SYSTEM},
        {"role": "user", "content": f"Tarefa: {resumo_da_tarefa}\n\nEntrada:\n{entrada}"}
    ]
    response = ollama.chat(model=MODEL, messages=messages)
    return response["message"]["content"]
```

```python
conteudo = "A fotossíntese é o processo pelo qual as plantas verdes, algas e algumas bactérias convertem a luz solar em energia química. Ela ocorre nos cloroplastos e utiliza dióxido de carbono e água como matérias-primas, produzindo glicose e oxigênio como produtos finais."
instrucoes_questoes = (
    "Gerar 5 QUESTÕES sobre o texto: "
    "- 2 de compreensão; 2 de aplicação; 1 de análise; "
    "incluir gabarito curto e, quando fizer sentido, rubrica de correção."
)
print(gerar(instrucoes_questoes, conteudo))
```

```python
import PyPDF2

def ler_pdf(caminho_arquivo):
    """
    Lê um arquivo PDF e retorna o texto de todas as páginas concatenado.
    """
    try:
        with open(caminho_arquivo, "rb") as f:
            reader = PyPDF2.PdfReader(f)
            texto = ""
            for page in reader.pages:
                texto += page.extract_text() or ""  # O 'or ""' evita problemas se a página não tiver texto
        return texto
    except FileNotFoundError:
        print(f"❌ ERRO: O arquivo '{caminho_arquivo}' não foi encontrado.")
        return ""
    except Exception as e:
        print(f"❌ ERRO ao ler o PDF: {e}")
        return ""

# Caminho para o seu PDF (substitua pelo nome real do seu arquivo)
caminho_do_pdf = "livro_tga.pdf"  # Lembre-se de colocar o PDF na mesma pasta do notebook

print(f"🔍 Tentando ler o PDF: {caminho_do_pdf}")

# Ler o PDF
texto_do_pdf = ler_pdf(caminho_do_pdf)

# Verificar se o texto foi extraído
if texto_do_pdf:
    print("\n✅ Texto extraído com sucesso!")
    print("Primeiros 500 caracteres:")
    print(texto_do_pdf[:500] + "...\n")
else:
    print("\n❌ O texto extraído está vazio. Isso pode significar que:")
    print("   - O PDF é uma imagem (escaneado) e não tem texto selecionável.")
    print("   - O PDF está protegido ou corrompido.")
    print("   - O caminho do arquivo está incorreto.")

# Se o texto foi extraído, gerar questões
if texto_do_pdf:
    # Definir instruções para gerar questões
    instrucoes_questoes = (
        "Gerar 5 QUESTÕES em português brasileiro sobre o texto: "
        "- 2 de compreensão; 2 de aplicação; 1 de análise; "
        "incluir gabarito curto e, quando fizer sentido, rubrica de correção."
    )

    print("🚀 Gerando questões a partir do PDF...")
    resposta = gerar(instrucoes_questoes, texto_do_pdf)
    print("\n📝 RESPOSTA GERADA:")
    print(resposta)
else:
    print("\n⚠️ Não foi possível gerar questões porque o texto do PDF está vazio.")
```
